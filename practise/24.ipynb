```python
import cv2
import time
```

```python
camera = cv2.VideoCapture(0)
```

```python
cv2.namedWindow('Neko\'s Camera', cv2.WINDOW_FREERATIO)

interval = 20
count = 0
```

```python
font = cv2.FONT_HERSHEY_SIMPLEX
fontScale = 1.5
fontColorGreen = (0, 255, 0)
fontThickness = 5
```

```python
# Hiển thị từng khung ảnh
while True:
    startTime = time.time()

    # Đọc 1 frame
    ret, frame = camera.read()

    # time.sleep(1 / 60)

    endTime = time.time()

    count += 1
    if count % interval == 0:
        cv2.imwrite("../Images/recorded/neko_frame1_{}.jpg".format(count), frame)

    fps = 0
    if endTime - startTime != 0: # Avoid division by zero error
        fps = 1/(endTime - startTime)

    cv2.putText(frame, "FPS: {:.2f}".format(fps), (10, 50), font, fontScale, fontColorGreen, fontThickness)

    # Thoát khi không thể đọc được frame
    if not ret:
        break

    # Hiển thị
    cv2.imshow("Neko's Camera", frame)

    if cv2.waitKey(10) == ord("q"):
        break
```

```python
camera.release()
cv2.destroyAllWindows()
```
